```python
import pandas as pd
import numpy as np
data = {'course' : ['math','phys','chem','biol'],
        'visa' : [80,70,60,np.nan],
        'final' : [68,90,np.nan,np.nan]
        }
df_lack = pd.DataFrame(data)
print(df_lack)
print("******************************************")
print(df_lack.isnull())  # hangi hücrelrein boş olduğunu true false olarak verir
print(df_lack.isnull().sum())   # hangi sütunda kaç tane boş sayısını verir.
print(df_lack.dropna())    # veri girilmemeiş (np.nan) olan satırların tamamını siler
print(df_lack.fillna(50))   # veri girilemmeişlere parantez içindeki değeri koyar. veri kaybetmemek için kullanılabilir.
```

```python
# (groupby)
import pandas as pd
infos = {
    'depart' : ['it','ik','serving','org','it'],
    'names' : ['ahmet','mehmet','ayşe','fatma','hayriye'],
    'payment' : [50000,45000,60000,55000,70000]
}
df = pd.DataFrame(infos)
avg_payment = df.groupby('depart')['payment'].mean()
print(avg_payment)
total_payment = df.groupby('depart')['payment'].sum()
print(total_payment)
count = df.groupby('depart')['payment'].count()
print(count)
max = df.groupby('depart')['payment'].max()
print(max)
```

```python
import pandas as pd

infos = {
    'depart': ['it', 'ik', 'serving', 'org', 'it'],
    'names': ['ahmet', 'mehmet', 'ayşe', 'fatma', 'hayriye'],
    'payment': [50000, 45000, 60000, 55000, 70000]
}
df = pd.DataFrame(infos)
print(df)
df['net_payment'] = df['payment']*1.10  # yeni sütun ekleme
print(df)
print("*************************************")
df.drop('net_payment',axis=1, inplace = True )   # Satır veya sütun silmek içindir. axis = 1 sütunu siler, axis = 0 satır siler, inplace = True yazarsak orjinal dataframede de siler olmazsa orjinalde silmez
print(df)
print("***************************************")
df_line1 = df.sort_values(by='payment')                     # sıralamak içindir, varsayılanı küçükten büyüğe sıralamaktır.(varsayılan)
df_line2 = df.sort_values(by='payment', ascending = False ) # ascending = False yazdığında büyükten küçüğe sıralar
print(df_line1)
print("------------------------------------------")
print(df_line2)
```

```python
import pandas as pd
veri ={
    'depart': ['fixing','it','ik','org'],
    'names': ['ali','veli','ayse','fatma'],
    'payment': [40000,50000,60000,80000]
}
df = pd.DataFrame(veri)
print(df)
df['tax']= df['payment']*0.20  # yeni sütun ekleme
print(df)
print("-------------------------------------------")
df.drop('tax', axis = 1, inplace = True) #sütunu silme, orjinalde de siliyor
print(df)
print("--------------------------------------------")
df_line11=df.sort_values(by ='payment')   #küçükten büyüğe sırlama
print(df_line11)
df_line21 = df.sort_values(by= 'payment', ascending = False)      # büyükten küçüğe sıralama
print(df_line21)
```
